```python
import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return np.exp(x)

def diferencia_finitas_centrada(f, x, h):
    return (f(x + h) - f(x - h)) / (2 * h)

x = 1

h = 0.1

derivada_aproximada = diferencia_finitas_centrada(f, x, h)

print(f"La derivada aproximada de f(x) = e^x en x = {x} con h = {h} es {derivada_aproximada}")

x_vals = np.linspace(0, 2, 100)
y_vals = f(x_vals)
y_prime_exacta = np.exp(x_vals)
plt.figure(figsize=(8, 6))
plt.plot(x_vals, y_vals, label="f(x) = e^x", color='blue')
plt.plot(x_vals, y_prime_exacta, label="Derivada exacta (f'(x) = e^x)", color='red', linestyle='dashed')
plt.scatter(x, f(x), color='blue')
plt.scatter(x, derivada_aproximada, color='red')
plt.title("Aproximación de la Derivada de f(x) = e^x en x = 1")
plt.xlabel("x")
plt.ylabel("f(x) y f'(x)")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def f(x, y):
    return x**2 - y**2

def gradiente_f(x, y):
    grad_x = 2 * x  # Derivada parcial de f respecto a x
    grad_y = -2 * y # Derivada parcial de f respecto a y
    return grad_x, grad_y

x_vals = np.linspace(-2, 2, 20)
y_vals = np.linspace(-2, 2, 20)
X, Y = np.meshgrid(x_vals, y_vals)

U, V = np.zeros(X.shape), np.zeros(Y.shape)
for i in range(X.shape[0]):
    for j in range(X.shape[1]):
        grad_x, grad_y = gradiente_f(X[i, j], Y[i, j])
        U[i, j] = grad_x
        V[i, j] = grad_y

plt.figure(figsize=(8, 6))
plt.quiver(X, Y, U, V, color='b', scale=20)
plt.title("Campo de Gradientes de f(x, y) = x^2 - y^2")
plt.xlabel("x")
plt.ylabel("y")
plt.xlim([-2, 2])
plt.ylim([-2, 2])
plt.grid(True)
plt.show()
```

```python
def f(x):
    return x**3 - 2*x + 1

def df(x):
    return 3*x**2 - 2

def newton_raphson(x0, tol=1e-6, max_iter=100):
    x_n = x0
    for i in range(max_iter):
        fx_n = f(x_n)
        dfx_n = df(x_n)

        if dfx_n == 0:
            print("La derivada es cero. No se puede continuar.")
            return None

        x_n1 = x_n - fx_n / dfx_n

        if abs(x_n1 - x_n) < tol:
            print(f"Convergió en {i+1} iteraciones.")
            return x_n1

        x_n = x_n1

    print("No convergió en el número máximo de iteraciones.")
    return x_n

x0 = 0.5

raiz = newton_raphson(x0)
print(f"La raíz aproximada es: {raiz}")
```
